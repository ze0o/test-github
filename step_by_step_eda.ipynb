# EDA Framework

For future datasets, remember this sequence:

**Understand → Validate → Clean → Describe → Visualize → Compare → Investigate → Explain → Recommend**

A data analyst should not stop at *"sales decreased."* Continue asking:

**Where? → When? → Which segment/product/customer? → What variables are associated with it? → Is it an anomaly or recurring pattern? → What should the business investigate or do next?**

## Step 3 — Understand structure and data types

**Ask:** Are numeric fields numeric? Is the date actually stored as a date?


```python
import pandas as pd

df = pd.read_csv("listings.csv")

date_columns = [
    "last_scraped",
    "price_quote_checkin_date",
    "price_quote_checkout_date",
    "calendar_last_scraped",
    "first_review",
    "last_review"
]

for column in df.columns:
    if column in date_columns:
        continue

    text_values = (
        df[column]
        .astype("string")
        .str.strip()
        .replace(["", "Unknown", "None", "nan"], pd.NA)
        .str.replace(",", "", regex=False)
        .str.replace("$", "", regex=False)
    )

    numeric_values = pd.to_numeric(text_values, errors="coerce")

    valid_values = text_values.notna()

    # Convert only if every non-empty value is numeric
    if valid_values.any() and numeric_values[valid_values].notna().all():
        if (numeric_values[valid_values] % 1 == 0).all():
            df[column] = numeric_values.astype("Int64")
        else:
            df[column] = numeric_values.astype("Float64")

df.info()
```

## Step 10 — Detect outliers using IQR

The common IQR rule defines potential outliers as values below:

`Q1 - 1.5 × IQR`

or above:

`Q3 + 1.5 × IQR`


```python
outlier_summary = []
outliers = {}

for col in numbers.columns:
    values = numbers[col].dropna()

    q1 = values.quantile(0.25)
    q3 = values.quantile(0.75)
    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    mask = (numbers[col] < lower_bound) | (numbers[col] > upper_bound)

    outliers[col] = df.loc[mask]

    outlier_summary.append({
        "Column": col,
        "Q1": q1,
        "Q3": q3,
        "IQR": iqr,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": mask.sum()
    })

outlier_summary = pd.DataFrame(outlier_summary)

outlier_summary.sort_values(
    by="Outlier Count",
    ascending=False
)
```

## Step 13 — Time-series EDA

**Ask:** what is trend over time? Are some months stronger than others?


```python
df["year_month"] = df["last_review"].dt.to_period("M")

monthly_trend = (
    df.groupby("year_month")["number_of_reviews"]
      .sum()
)

print(monthly_trend)
```

```python
df["year_month"] = df["last_review"].dt.to_period("M")

monthly_trend = (
    df.groupby("year_month")["number_of_reviews"]
      .sum()
)

print(monthly_trend)
```

```python
monthly_revenue = (
    df.groupby("year_month")["estimated_revenue_l365d"]
      .sum()
)


monthly_price = (
    df.groupby("year_month")["price"]
      .mean()
      .round(2)
)

df["year"] = df["last_review"].dt.year

yearly_reviews = (
    df.groupby("year")["number_of_reviews"]
      .mean()
      .round(2)
)

print("\nAverage reviews by year:")
print(yearly_reviews)
print("\nEstimated revenue by month:")
print(monthly_revenue)
print("\nAverage price by month:")
print(monthly_price)
```

```python
monthly_listings = (
    df.groupby("year_month")["id"]
      .count()
)



df["month"] = df["last_review"].dt.month

month_strength = (
    df.groupby("month")["number_of_reviews"]
      .mean()
      .round(2)
)

print("\nNumber of listings reviewed each month:")
print(monthly_listings)
print("\nAverage reviews by month of year:")
print(month_strength)
```

## Step 14 — Find the best and worst performers

A useful EDA habit is to inspect both extremes rather than only averages.


```python
col = [
    "id",
    "name",
    "room_type",
    "price",
    "estimated_revenue_l365d"
]

#by revenue
best_revenue = df.sort_values(
    "estimated_revenue_l365d",
    ascending=False
).head(10)

worst_revenue = df.sort_values(
    "estimated_revenue_l365d",
    ascending=True
).head(10)

print("TOP 10 — Highest Revenue")
print(best_revenue[col])

print("\nBOTTOM 10 — Lowest Revenue")
print(worst_revenue[col])


#by rating
best_rating = df.sort_values(
    "review_scores_rating",
    ascending=False
).head(10)

worst_rating = df.sort_values(
    "review_scores_rating",
    ascending=True
).head(10)

print("\nTOP 10 — Highest Rating")
print(best_rating[col])

print("\nBOTTOM 10 — Lowest Rating")
print(worst_rating[col])
```
